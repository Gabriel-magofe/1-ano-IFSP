## Lista de exercício cálculo de acurácia

### Questão 1. Filtro de Spam

- **Dados:**

  - VN = 130
  - VP = 40
  - FN = 20
  - FP = 10
  - *Total = 200*

- **Cálculo da Acurácia:**
 
  $$\text{Acurácia} = \frac{VP + VN}{Total} = \frac{40 + 130}{200} = \frac{170}{200} = 0,85$$

- **Resultado:** 85%
